## Visualizing the data

```python
flood_counts = df['FLOODS'].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(flood_counts, labels=flood_counts.index, autopct='%1.1f%%', colors=['red', 'lightgreen'], startangle=140)
plt.title('Proportion of Flood vs No Flood Years in Kerala (1901-2018)')
plt.axis('equal')  # Equal aspect ratio ensures the pie chart is circular
plt.tight_layout()
plt.show()
```

### Flood faced by Kerala between 1901-2018 in percentage

```python
monthly_columns = ['JAN', 'FEB', 'MAR', 'APR', 'MAY', 'JUN', 'JUL', 'AUG', 'SEP', 'OCT', 'NOV', 'DEC']
monthly_avg = df[monthly_columns].mean()

plt.figure(figsize=(10, 6))
monthly_avg.plot(kind='bar', color='skyblue')
plt.title('Average Monthly Rainfall in Kerala (1901-2018)')
plt.ylabel('Rainfall (mm)')
plt.xlabel('Months')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

### June,July months has been got the most rainfall in Kerala 

```python
months = ['JAN', 'FEB', 'MAR', 'APR', 'MAY', 'JUN', 
          'JUL', 'AUG', 'SEP', 'OCT', 'NOV', 'DEC']
plt.figure(figsize=(14, 8))
sns.boxplot(data=df[months], palette="Blues")
plt.title('Monthly Rainfall Distribution in Kerala (1901-2018)', fontsize=16)
plt.xlabel('Month', fontsize=14)
plt.ylabel('Rainfall (mm)', fontsize=14)
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

### The most consistant rainfall happens in July month also June month is also nearly to July month and extreme rainfall happens is also in july and june month

```python
months = ['JAN', 'FEB', 'MAR', 'APR', 'MAY', 'JUN', 
          'JUL', 'AUG', 'SEP', 'OCT', 'NOV', 'DEC']
heatmap_data = df[['YEAR'] + months].copy()
heatmap_data.set_index('YEAR', inplace=True)
plt.figure(figsize=(14, 10))
sns.heatmap(heatmap_data, cmap="YlGnBu", linewidths=0.5, linecolor='gray')
plt.title('Monthly Rainfall Heatmap in Kerala (1901–2018)', fontsize=16)
plt.xlabel('Month',fontsize=20)
plt.ylabel('Year',fontsize=20)
plt.tight_layout()
plt.show()
```

### The june and july months has been got the most rainfall in every year

```python
sns.barplot(x=df['FLOODS'],y=df[' ANNUAL RAINFALL'])
plt.title('Annual rainfall and floods')
plt.show()
```

### Annualy rainfall and chance of happening flood in Kerala

```python
maxs=df.loc[df[' ANNUAL RAINFALL'].idxmax()]
max_year=maxs['YEAR']
max_rain=maxs[' ANNUAL RAINFALL']
plt.figure(figsize=(12, 6))
plt.plot(df['YEAR'], df[' ANNUAL RAINFALL'], label='Annual Rainfall', color='gray')
plt.scatter(max_year, max_rain, color='blue', s=100, zorder=5, label=f'Max Rainfall: {max_year} ({max_rain:.1f} mm)')
plt.title('Annualy rainfall in year base')
plt.xlabel('Year')
plt.ylabel('Annual rainfall')
plt.show()
```

### The most rainfall happens in 2018 that is the plot is showing

```python
months=['JAN', 'FEB', 'MAR', 'APR', 'MAY', 'JUN', 
          'JUL', 'AUG', 'SEP', 'OCT', 'NOV', 'DEC']
rain=df[df['YEAR']==2018].squeeze()
plt.bar(months,rain[months],color='red')
plt.title('Monthly Rainfall in Kerala (2018)', fontsize=16)
plt.xlabel('Month of 2018')
plt.ylabel('Rainfall (mm)')
plt.show()
```
